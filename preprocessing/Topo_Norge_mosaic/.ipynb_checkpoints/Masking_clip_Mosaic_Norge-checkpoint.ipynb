```python
#####Bad Pixels not masked###

import os
import time
import arcpy
from arcpy.sa import ExtractByMask

# -------------------------------------------------
# USER SETTINGS
# -------------------------------------------------
in_folder    = r"E:\Mosaic_Norge\Test"  # mosaics to be masked
mask_raster  = r"E:\Mosaic_Norge\snap_raster\dtm_mask_srast_fixed.tif"

# ✅ requested output folder
out_folder   = r"E:\Mosaic_Norge\Final_rasts"
out_suffix   = "_masked"

TARGET_CELLSIZE = 10  # meters

build_stats     = True
build_pyramids  = True
overwrite       = True
# -------------------------------------------------

arcpy.env.overwriteOutput = overwrite
os.makedirs(out_folder, exist_ok=True)

def log(msg):
    print(msg)
    try:
        arcpy.AddMessage(msg)
    except:
        pass

def format_eta(seconds):
    seconds = max(0, int(seconds))
    h = seconds // 3600
    m = (seconds % 3600) // 60
    s = seconds % 60
    if h > 0:
        return f"{h}h {m}m {s}s"
    if m > 0:
        return f"{m}m {s}s"
    return f"{s}s"

def progress_bar(i, n, width=28):
    if n <= 0:
        return ""
    frac = i / n
    filled = int(round(frac * width))
    bar = "█" * filled + "░" * (width - filled)
    return f"[{bar}] {i}/{n} ({int(frac*100)}%)"

def signature(r):
    d = arcpy.Describe(r)
    ext = d.extent
    return (
        d.spatialReference.factoryCode if d.spatialReference else None,
        round(d.meanCellWidth, 6),
        round(d.meanCellHeight, 6),
        round(ext.XMin, 6),
        round(ext.YMax, 6),
        int(d.width),
        int(d.height)
    )

# -------------------------------------------------
# Checks + environment lock
# -------------------------------------------------
if not arcpy.Exists(mask_raster):
    raise RuntimeError(f"Mask raster not found: {mask_raster}")

try:
    arcpy.CheckOutExtension("Spatial")
except:
    raise RuntimeError("Spatial Analyst extension is not available. ExtractByMask requires it.")

mask_sig = signature(mask_raster)
mask_desc = arcpy.Describe(mask_raster)
mask_ext  = mask_desc.extent
mask_sr   = mask_desc.spatialReference
mask_cell = mask_desc.meanCellWidth

if round(mask_cell, 6) != float(TARGET_CELLSIZE):
    log(f"WARNING: mask cellsize is {mask_cell}, but you requested {TARGET_CELLSIZE}. "
        f"Outputs will be forced to {TARGET_CELLSIZE}m.")

# 🔒 Lock to mask grid
arcpy.env.snapRaster = mask_raster
arcpy.env.mask = mask_raster
arcpy.env.extent = mask_ext
arcpy.env.cellSize = TARGET_CELLSIZE
arcpy.env.outputCoordinateSystem = mask_sr

# Collect inputs
tifs = [f for f in os.listdir(in_folder) if f.lower().endswith((".tif", ".tiff"))]
if not tifs:
    raise RuntimeError(f"No .tif/.tiff files found in: {in_folder}")

tifs = sorted(tifs)

log(f"Found {len(tifs)} rasters in: {in_folder}")
log(f"Mask/snap raster: {mask_raster}")
log(f"Output folder   : {out_folder}")
log(f"Forced cell size: {TARGET_CELLSIZE} m")
log(f"Mask signature  : {mask_sig}")

# -------------------------------------------------
# Run
# -------------------------------------------------
t0_global = time.time()
times = []

failed = []
mismatch = []

for idx, fn in enumerate(tifs, start=1):
    in_r = os.path.join(in_folder, fn)
    base, _ = os.path.splitext(fn)
    out_r = os.path.join(out_folder, f"{base}{out_suffix}.tif")

    # If overwrite is False and output exists
    if arcpy.Exists(out_r) and not overwrite:
        dt = 0.0
        # still verify alignment if it exists
        try:
            out_sig = signature(out_r)
            if out_sig != mask_sig:
                mismatch.append((out_r, out_sig))
        except:
            mismatch.append((out_r, "could not read signature"))
        avg = (sum(times) / len(times)) if times else 0
        eta = format_eta((len(tifs) - idx) * avg) if avg > 0 else "?"
        log(f"{progress_bar(idx, len(tifs))} EXISTS (skip) | ETA: {eta} | {fn}")
        continue

    t1 = time.time()
    try:
        masked = ExtractByMask(in_r, mask_raster)
        masked.save(out_r)

        if build_stats:
            arcpy.management.CalculateStatistics(out_r, 1, 1, "OVERWRITE")
        if build_pyramids:
            arcpy.management.BuildPyramids(out_r)

        # ✅ Check alignment/dimensions match mask
        out_sig = signature(out_r)
        if out_sig != mask_sig:
            mismatch.append((out_r, out_sig))

    except Exception as e:
        failed.append((in_r, str(e)))
        log(f"{progress_bar(idx, len(tifs))} FAILED | {fn}\n    {e}")
        continue

    dt = time.time() - t1
    times.append(dt)
    avg = sum(times) / len(times)
    eta = format_eta((len(tifs) - idx) * avg)

    # brief per-file log
    ok_txt = "OK" if (out_sig == mask_sig) else "ALIGNMENT MISMATCH"
    log(f"{progress_bar(idx, len(tifs))} {ok_txt} | {dt:.1f}s | ETA: {eta} | {fn}")

# -------------------------------------------------
# Summary
# -------------------------------------------------
total = time.time() - t0_global
log("\n" + "=" * 70)
log("DONE")
log(f"Total runtime: {total/60:.2f} minutes")
log(f"Outputs in   : {out_folder}")
log(f"Mask signature: {mask_sig}")

if mismatch:
    log("\nWARNING: Some outputs do NOT exactly match the mask grid signature (SR, cell, XMin, YMax, width, height):")
    for p, sig in mismatch[:20]:
        log(f" - {p} -> {sig}")
    if len(mismatch) > 20:
        log(f" ... and {len(mismatch)-20} more")
else:
    log("\nAll masked mosaics match the mask alignment + dimensions exactly ✅")

if failed:
    log("\nFAILED files:")
    for p, err in failed[:20]:
        log(f" - {p}\n   {err}")
    if len(failed) > 20:
        log(f" ... and {len(failed)-20} more")
log("=" * 70)
```

```python
#####Eliminating bad pixels in Land + labelling rasters that have bad pixels as _ch + THRESHOLD EDITED#####

import os
import time
import arcpy
from arcpy.sa import ExtractByMask, SetNull, Raster

# -------------------------------------------------
# USER SETTINGS
# -------------------------------------------------
in_folder    = r"F:\Topodata_basins\Mosaic_Norge\Test"
mask_raster  = r"F:\Topodata_basins\Mosaic_Norge\snap_raster\dtm_mask_srast_fixed.tif"

out_folder   = r"F:\Topodata_basins\Mosaic_Norge\Final_rasts\Ready"
out_suffix   = "_masked"

TARGET_CELLSIZE = 10

build_stats     = True
build_pyramids  = True
overwrite       = True

# -------------------------------------------------
# Threshold rules (substring match in tif name)
# -------------------------------------------------
THRESH_RULES = {
    # tri & spi: values > 100 are bad
    "tri": {"type": "upper", "upper": 100},
    "spi": {"type": "upper", "upper": 100},

    # slope & dtm: values > 1e9 are bad
    "slope": {"type": "upper", "upper": 1e9},
    "dtm":   {"type": "upper", "upper": 1e9},

    # curvatures: values outside [-15, 15] are bad
    "curv_plan": {"type": "range", "lower": -15, "upper": 15},
    "curv_prof": {"type": "range", "lower": -15, "upper": 15},
    "curv_tot":  {"type": "range", "lower": -15, "upper": 15},

    # aspect: values < 0 OR values > 1e9 are bad
    "aspect": {"type": "range", "lower": 0, "upper": 1e9},
}

# Fallback if no rule matched (safe default)
DEFAULT_RULE = {"type": "upper", "upper": 1e9}
# -------------------------------------------------

arcpy.env.overwriteOutput = overwrite
os.makedirs(out_folder, exist_ok=True)

def log(msg):
    print(msg)
    try:
        arcpy.AddMessage(msg)
    except:
        pass

def format_eta(seconds):
    seconds = max(0, int(seconds))
    h = seconds // 3600
    m = (seconds % 3600) // 60
    s = seconds % 60
    if h > 0:
        return f"{h}h {m}m {s}s"
    if m > 0:
        return f"{m}m {s}s"
    return f"{s}s"

def progress_bar(i, n, width=28):
    frac = i / n
    filled = int(round(frac * width))
    bar = "█" * filled + "░" * (width - filled)
    return f"[{bar}] {i}/{n} ({int(frac*100)}%)"

def signature(r):
    d = arcpy.Describe(r)
    ext = d.extent
    return (
        d.spatialReference.factoryCode if d.spatialReference else None,
        round(d.meanCellWidth, 6),
        round(d.meanCellHeight, 6),
        round(ext.XMin, 6),
        round(ext.YMax, 6),
        int(d.width),
        int(d.height)
    )

def get_rule_for_filename(filename):
    """Pick the first matching rule by substring; otherwise DEFAULT_RULE."""
    name = filename.lower()
    for key, rule in THRESH_RULES.items():
        if key in name:
            return key, rule
    return None, DEFAULT_RULE

def bad_mask(raster_obj, rule):
    """
    Returns a boolean raster where True indicates 'bad' pixels, based on rule.
    - upper: bad if value > upper
    - lower: bad if value < lower
    - range: bad if value < lower OR value > upper
    """
    rtype = rule.get("type")
    if rtype == "upper":
        return raster_obj > rule["upper"]
    elif rtype == "lower":
        return raster_obj < rule["lower"]
    elif rtype == "range":
        return (raster_obj < rule["lower"]) | (raster_obj > rule["upper"])
    else:
        raise ValueError(f"Unknown rule type: {rtype}")

# -------------------------------------------------
# Environment lock
# -------------------------------------------------
arcpy.env.snapRaster = mask_raster
arcpy.env.mask = mask_raster
arcpy.env.extent = arcpy.Describe(mask_raster).extent
arcpy.env.cellSize = TARGET_CELLSIZE
arcpy.env.outputCoordinateSystem = arcpy.Describe(mask_raster).spatialReference

# Collect inputs
tifs = sorted(f for f in os.listdir(in_folder) if f.lower().endswith(".tif"))

log(f"Found {len(tifs)} rasters")
log("Bad-pixel rules (by filename substring):")
log("  tri/spi      : value > 100               -> NoData")
log("  slope/dtm    : value > 1e9               -> NoData")
log("  curv_*       : value outside [-15, 15]   -> NoData")
log("  aspect       : value < 0 or value > 1e9  -> NoData")
log("  default      : value > 1e9               -> NoData (if no name match)")

# -------------------------------------------------
# Run
# -------------------------------------------------
t0_global = time.time()
times = []

for idx, fn in enumerate(tifs, start=1):
    in_r = os.path.join(in_folder, fn)
    base, _ = os.path.splitext(fn)

    t1 = time.time()

    try:
        rule_key, rule = get_rule_for_filename(fn)

        r_in = Raster(in_r)

        # ---------- detect bad pixels (rule-specific) ----------
        bad_test = bad_mask(r_in, rule)
        has_bad = float(arcpy.management.GetRasterProperties(
            bad_test, "MAXIMUM"
        )[0]) > 0
        # ------------------------------------------------------

        # build output name
        suffix = out_suffix + ("_ch" if has_bad else "")
        out_r = os.path.join(out_folder, f"{base}{suffix}.tif")

        masked = ExtractByMask(in_r, mask_raster)

        if has_bad:
            # ---------- remove bad pixels (rule-specific) ----------
            masked_r = Raster(masked)
            masked = SetNull(bad_mask(masked_r, rule), masked_r)
            # ------------------------------------------------------

        masked.save(out_r)

        if build_stats:
            arcpy.management.CalculateStatistics(out_r, 1, 1, "OVERWRITE")
        if build_pyramids:
            arcpy.management.BuildPyramids(out_r)

        out_sig = signature(out_r)

    except Exception as e:
        log(f"{progress_bar(idx, len(tifs))} FAILED | {fn}\n    {e}")
        continue

    dt = time.time() - t1
    times.append(dt)
    avg = sum(times) / len(times)
    eta = format_eta((len(tifs) - idx) * avg)

    tag = "CH" if has_bad else "OK"
    rule_tag = rule_key if rule_key else "default"
    log(f"{progress_bar(idx, len(tifs))} {tag} | {dt:.1f}s | ETA: {eta} | rule={rule_tag} | {fn}")

# -------------------------------------------------
# Summary
# -------------------------------------------------
total = time.time() - t0_global
log("\n" + "=" * 70)
log("DONE")
log(f"Total runtime: {total/60:.2f} minutes")
log(f"Outputs in   : {out_folder}")
log("=" * 70)

```

```python
# ================================================================
# BASIN TILES -> CLIP -> NATIONAL MOSAIC -> ALIGN MASK -> CLEAN -> COG
# ================================================================

import os
import time
import arcpy
from arcpy.sa import Raster, SetNull, IsNull

# -------------------------------------------------
# USER SETTINGS
# -------------------------------------------------
root_dir = r"F:\Topodata_basins\Data10m"

basins_fc = r"F:\Topodata_basins\Polygon_basins\Gruppering.shp"
basin_id_field = "RegineEnhe"

template_mask = r"E:\mask\Mask_land_Kalk_cog.tif"

raster_names = [
    "aspect",
    "curv_plan",
    "curv_prof",
    "curv_tot",
    "slope",
    "dtm",
    "tri",
    "spi"
]

out_root = r"F:\Mosaic_Norge\Final_COGs"
tmp_root = r"F:\Mosaic_Norge\_tmp_processing"

os.makedirs(out_root, exist_ok=True)
os.makedirs(tmp_root, exist_ok=True)

OVERWRITE = True
KEEP_TEMP_CLIPS = False

pixel_type = "32_BIT_FLOAT"
mosaic_method = "LAST"

BAD_PIXEL_ABS = 1e15

# -------------------------------------------------
# BAD PIXEL RULES
# -------------------------------------------------
THRESH_RULES = {
    "tri":       {"type": "upper", "upper": 100},
    "spi":       {"type": "upper", "upper": 100},
    "slope":     {"type": "range", "lower": 0, "upper": 90},
    "dtm":       {"type": "upper", "upper": 1e9},
    "curv_plan": {"type": "range", "lower": -15, "upper": 15},
    "curv_prof": {"type": "range", "lower": -15, "upper": 15},
    "curv_tot":  {"type": "range", "lower": -15, "upper": 15},
    "aspect":    {"type": "range", "lower": 0, "upper": 360},
}

# -------------------------------------------------
# SETUP
# -------------------------------------------------
arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

mask_r = Raster(template_mask)
mask_desc = arcpy.Describe(template_mask)
mask_sr = mask_desc.spatialReference
mask_ext = mask_desc.extent
mask_cell = float(mask_r.meanCellWidth)

arcpy.env.snapRaster = template_mask
arcpy.env.extent = mask_ext
arcpy.env.cellSize = mask_cell
arcpy.env.outputCoordinateSystem = mask_sr
arcpy.env.compression = "LZW"
arcpy.env.pyramid = "NONE"
arcpy.env.rasterStatistics = "NONE"

# -------------------------------------------------
# HELPERS
# -------------------------------------------------
def log(msg):
    print(msg)
    try:
        arcpy.AddMessage(msg)
    except Exception:
        pass

def fmt_time(seconds):
    seconds = int(round(max(0, seconds)))
    h = seconds // 3600
    m = (seconds % 3600) // 60
    s = seconds % 60
    return f"{h}:{m:02d}:{s:02d}" if h else f"{m}:{s:02d}"

def safe_name(x):
    return (
        str(x).replace(" ", "_")
              .replace("/", "_")
              .replace("\\", "_")
              .replace(":", "_")
              .replace("(", "")
              .replace(")", "")
    )

def delete_if_exists(path):
    try:
        if arcpy.Exists(path):
            arcpy.management.Delete(path)
    except Exception:
        pass

def find_raster_in_folder(folder, raster_name):
    """
    Searches recursively inside one basin folder.
    Priority:
    1. exact filename match: curv_plan.tif
    2. filename contains raster_name
    """
    exact = []
    partial = []

    for dirpath, _, filenames in os.walk(folder):
        for fn in filenames:
            name, ext = os.path.splitext(fn)
            if ext.lower() not in [".tif", ".tiff"]:
                continue

            full = os.path.join(dirpath, fn)

            if name.lower() == raster_name.lower():
                exact.append(full)
            elif raster_name.lower() in name.lower():
                partial.append(full)

    if exact:
        return exact[0]

    if partial:
        return partial[0]

    return None

def get_basin_ids():
    ids = []
    with arcpy.da.SearchCursor(basins_fc, [basin_id_field]) as cur:
        for (bid,) in cur:
            if bid is not None:
                ids.append(str(bid).strip())
    return sorted(set(ids))

def bad_mask(r, rule):
    extreme = (r >= BAD_PIXEL_ABS) | (r <= -BAD_PIXEL_ABS)

    if rule["type"] == "upper":
        return extreme | (r > rule["upper"])

    if rule["type"] == "range":
        return extreme | (r < rule["lower"]) | (r > rule["upper"])

    return extreme

def get_rule(raster_name):
    lname = raster_name.lower()
    for key, rule in THRESH_RULES.items():
        if key in lname:
            return rule
    return {"type": "upper", "upper": BAD_PIXEL_ABS}

def check_grid(out_raster):
    r = Raster(out_raster)

    ok = (
        r.width == mask_r.width and
        r.height == mask_r.height and
        abs(r.meanCellWidth - mask_r.meanCellWidth) < 1e-9 and
        abs(r.meanCellHeight - mask_r.meanCellHeight) < 1e-9 and
        abs(r.extent.XMin - mask_r.extent.XMin) < 1e-6 and
        abs(r.extent.YMin - mask_r.extent.YMin) < 1e-6 and
        abs(r.extent.XMax - mask_r.extent.XMax) < 1e-6 and
        abs(r.extent.YMax - mask_r.extent.YMax) < 1e-6
    )

    log(f"Grid check: {'OK' if ok else 'WARNING'} | {r.width} x {r.height}")
    return ok

def minmax(raster_path):
    try:
        mn = arcpy.management.GetRasterProperties(raster_path, "MINIMUM").getOutput(0)
        mx = arcpy.management.GetRasterProperties(raster_path, "MAXIMUM").getOutput(0)
        return mn, mx
    except Exception:
        return "NA", "NA"

# -------------------------------------------------
# CHECKS
# -------------------------------------------------
if not arcpy.Exists(basins_fc):
    raise RuntimeError(f"Basins shapefile not found: {basins_fc}")

if not arcpy.Exists(template_mask):
    raise RuntimeError(f"Template mask not found: {template_mask}")

basin_ids = get_basin_ids()

log(f"Found {len(basin_ids)} basin polygons.")
log(f"Template mask: {template_mask}")
log(f"Variables: {', '.join(raster_names)}")

# -------------------------------------------------
# MAIN LOOP
# -------------------------------------------------
global_start = time.time()

for v_idx, raster_name in enumerate(raster_names, start=1):

    var_start = time.time()

    log("\n" + "=" * 80)
    log(f"[{v_idx}/{len(raster_names)}] PROCESSING: {raster_name}")
    log("=" * 80)

    out_base = safe_name(raster_name)
    out_ok = os.path.join(out_root, f"{out_base}.tif")
    out_ch = os.path.join(out_root, f"{out_base}_ch.tif")

    if not OVERWRITE and (arcpy.Exists(out_ok) or arcpy.Exists(out_ch)):
        log(f"Output exists. Skipping: {out_ch if arcpy.Exists(out_ch) else out_ok}")
        continue

    delete_if_exists(out_ok)
    delete_if_exists(out_ch)

    tmp_clip_dir = os.path.join(tmp_root, f"clips_{out_base}")
    os.makedirs(tmp_clip_dir, exist_ok=True)

    tmp_mosaic = os.path.join(tmp_root, f"{out_base}_mosaic.tif")
    tmp_aligned = os.path.join(tmp_root, f"{out_base}_aligned.tif")
    tmp_clean = os.path.join(tmp_root, f"{out_base}_clean.tif")

    delete_if_exists(tmp_mosaic)
    delete_if_exists(tmp_aligned)
    delete_if_exists(tmp_clean)

    rule = get_rule(raster_name)

    basins_lyr = f"basins_lyr_{out_base}"
    delete_if_exists(basins_lyr)
    arcpy.management.MakeFeatureLayer(basins_fc, basins_lyr)

    clipped_rasters = []
    clip_times = []

    try:
        # -----------------------------------------------------
        # 1. FIND EACH TILE AND CLIP BY BASIN POLYGON
        # -----------------------------------------------------
        log("Step 1/5: finding basin tiles and clipping...")

        for i, bid in enumerate(basin_ids, start=1):

            basin_folder = os.path.join(root_dir, bid)

            if not os.path.isdir(basin_folder):
                continue

            in_raster = find_raster_in_folder(basin_folder, raster_name)

            if in_raster is None:
                continue

            out_clip = os.path.join(tmp_clip_dir, f"{safe_name(bid)}_{out_base}.tif")
            delete_if_exists(out_clip)

            where = f"{arcpy.AddFieldDelimiters(basins_fc, basin_id_field)} = '{bid}'"
            arcpy.management.SelectLayerByAttribute(basins_lyr, "NEW_SELECTION", where)

            if int(arcpy.management.GetCount(basins_lyr).getOutput(0)) != 1:
                continue

            t0 = time.time()

            arcpy.management.Clip(
                in_raster=in_raster,
                rectangle="#",
                out_raster=out_clip,
                in_template_dataset=basins_lyr,
                nodata_value="#",
                clipping_geometry="ClippingGeometry",
                maintain_clipping_extent="NO_MAINTAIN_EXTENT"
            )

            clipped_rasters.append(out_clip)

            dt = time.time() - t0
            clip_times.append(dt)

            avg = sum(clip_times) / len(clip_times)
            eta = avg * (len(basin_ids) - i)

            log(f"  [{i}/{len(basin_ids)}] clipped {bid} | {fmt_time(dt)} | ETA {fmt_time(eta)}")

        arcpy.management.SelectLayerByAttribute(basins_lyr, "CLEAR_SELECTION")

        if not clipped_rasters:
            log(f"No clipped rasters found for {raster_name}. Check root_dir and filenames.")
            continue

        # -----------------------------------------------------
        # 2. MOSAIC ALL CLIPPED BASIN TILES
        # -----------------------------------------------------
        log(f"Step 2/5: mosaicking {len(clipped_rasters)} clipped tiles...")

        d0 = arcpy.Describe(clipped_rasters[0])
        src_sr = d0.spatialReference
        src_cell = d0.meanCellWidth
        src_bands = getattr(d0, "bandCount", 1)

        arcpy.management.MosaicToNewRaster(
            input_rasters=";".join(clipped_rasters),
            output_location=tmp_root,
            raster_dataset_name_with_extension=os.path.basename(tmp_mosaic),
            coordinate_system_for_the_raster=src_sr,
            pixel_type=pixel_type,
            cellsize=src_cell,
            number_of_bands=src_bands,
            mosaic_method=mosaic_method,
            mosaic_colormap_mode="FIRST"
        )

        # -----------------------------------------------------
        # 3. PROJECT/RESAMPLE ONCE TO FINAL MASK GRID
        # -----------------------------------------------------
        log("Step 3/5: projecting and aligning to final mask grid...")

        arcpy.management.ProjectRaster(
            in_raster=tmp_mosaic,
            out_raster=tmp_aligned,
            out_coor_system=mask_sr,
            resampling_type="BILINEAR",
            cell_size=mask_cell
        )

        # -----------------------------------------------------
        # 4. LAND MASK + BAD PIXEL REMOVAL
        # -----------------------------------------------------
        log("Step 4/5: masking land and removing bad pixels...")

        r = Raster(tmp_aligned)

        r_clean = SetNull(
            IsNull(mask_r) | bad_mask(r, rule),
            r
        )

        r_clean.save(tmp_clean)

        mn, mx = minmax(tmp_clean)
        log(f"Clean range: {mn} to {mx}")

        has_bad = False
        try:
            has_bad_test = bad_mask(Raster(tmp_aligned), rule)
            has_bad = float(arcpy.management.GetRasterProperties(has_bad_test, "MAXIMUM").getOutput(0)) > 0
        except Exception:
            has_bad = True

        out_final = out_ch if has_bad else out_ok

        # -----------------------------------------------------
        # 5. FINAL COG
        # -----------------------------------------------------
        log("Step 5/5: writing final Cloud Optimized GeoTIFF...")

        delete_if_exists(out_final)

        arcpy.management.CopyRaster(
            in_raster=tmp_clean,
            out_rasterdataset=out_final,
            pixel_type=pixel_type,
            format="COG"
        )

        arcpy.management.CalculateStatistics(out_final)

        check_grid(out_final)

        mn2, mx2 = minmax(out_final)
        log(f"Final COG range: {mn2} to {mx2}")

        log(f"DONE: {out_final}")
        log(f"Variable time: {fmt_time(time.time() - var_start)}")

    except Exception as e:
        log(f"ERROR processing {raster_name}: {e}")
        arcpy.AddWarning(f"ERROR processing {raster_name}: {e}")

    finally:
        delete_if_exists(basins_lyr)
        delete_if_exists(tmp_mosaic)
        delete_if_exists(tmp_aligned)
        delete_if_exists(tmp_clean)

        if not KEEP_TEMP_CLIPS:
            for rr in clipped_rasters:
                delete_if_exists(rr)
            try:
                os.rmdir(tmp_clip_dir)
            except Exception:
                pass

# -------------------------------------------------
# FINISH
# -------------------------------------------------
log("\n" + "=" * 80)
log(f"ALL DONE. Total runtime: {fmt_time(time.time() - global_start)}")
log(f"Outputs in: {out_root}")
log("=" * 80)

arcpy.CheckInExtension("Spatial")
```
