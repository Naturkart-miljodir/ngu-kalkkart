```python
###Rasterize category maps

# -*- coding: utf-8 -*-
import os
import time
import arcpy

# -----------------------
# TOGGLES
# -----------------------
DO_MARINE = False
DO_LANDUSE = True

DO_GEOLOGY_LOGCA = False
DO_GEOLOGY_CA_COE = False

DO_QUATERNARY = False

OVERWRITE_EXISTING = False  # False = skip if output exists

# -----------------------
# INPUTS
# -----------------------
MARINE_SHP = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Marine_limit\Marine_limit_to_raster.shp"

LANDUSE_SHP = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Landuse\Norway_corine\Norway_landuse.shp"

# IMPORTANT: shapefile must include .shp
GEOLOGY_CA_FC = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Berggrunns_Kalk_feb2026_clip.shp"

QUATERNARY_SHP = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Quaternary\LosmassteFlate_Lokal-Regional_forenklet_feb2026.shp"

MASK_RASTER = r"E:\mask\Mask_land_Kalk_cog.tif"

OUT_DIR = r"G:\Geology_rasters_2026\Landuse_selected"

# -----------------------
# FIELDS
# -----------------------
MARINE_FIELD = "ANNETKVTEM"
LANDUSE_FIELD = "Code_18"

GEOLOGY_LOGCA_FIELD = "LogCa_ICP_"
GEOLOGY_CA_COE_FIELD = "Ca_ICP_Coe"

QUATERNARY_FIELD = "ForenkletK"

# -----------------------
# OUTPUTS
# -----------------------
OUT_MARINE_COG = os.path.join(OUT_DIR, "marine_limit_ANNETKVTEM_cog.tif")
OUT_LANDUSE_COG = os.path.join(OUT_DIR, "landuse_Code_18_cog.tif")

OUT_GEOLOGY_LOGCA_COG = os.path.join(OUT_DIR, "geology_LogCa_ICP_cog.tif")
OUT_GEOLOGY_CA_COE_COG = os.path.join(OUT_DIR, "geology_Ca_ICP_Coe_cog.tif")

OUT_QUATERNARY_COG = os.path.join(OUT_DIR, "quaternary_ForenkletK_cog.tif")


# -----------------------
# HELPERS
# -----------------------
def ts():
    return time.strftime("%Y-%m-%d %H:%M:%S")

def msg(s):
    arcpy.AddMessage(f"[{ts()}] {s}")
    print(f"[{ts()}] {s}")

def list_fields(fc):
    return [f.name for f in arcpy.ListFields(fc)]

def require_field(fc, field_name):
    if field_name not in list_fields(fc):
        msg(f"Fields in {os.path.basename(fc)}:")
        for fn in list_fields(fc):
            msg(f"  - {fn}")
        raise ValueError(f"Field '{field_name}' not found in: {fc}")

def safe_delete(path):
    try:
        if arcpy.Exists(path):
            arcpy.management.Delete(path)
        elif os.path.exists(path):
            os.remove(path)
    except Exception:
        pass

def should_skip(out_path):
    return (not OVERWRITE_EXISTING) and os.path.exists(out_path)

def stage_geotiff(in_raster, stage_tif):
    safe_delete(stage_tif)
    arcpy.management.CopyRaster(
        in_raster=in_raster,
        out_rasterdataset=stage_tif,
        format="TIFF"
    )

def to_cog(in_raster, out_cog, stage_name):

    if should_skip(out_cog):
        msg(f"Skipping (already exists): {out_cog}")
        return

    if not os.path.isdir(os.path.dirname(out_cog)):
        os.makedirs(os.path.dirname(out_cog))

    if OVERWRITE_EXISTING:
        safe_delete(out_cog)

    scratch_folder = arcpy.env.scratchFolder or os.environ.get("TEMP", r"C:\Temp")
    stage_tif = os.path.join(scratch_folder, f"{stage_name}_stage.tif")

    msg(f"Staging GeoTIFF: {stage_tif}")
    stage_geotiff(in_raster, stage_tif)

    msg(f"Writing COG: {out_cog}")
    safe_delete(out_cog)

    arcpy.management.CopyRaster(
        in_raster=stage_tif,
        out_rasterdataset=out_cog,
        format="COG"
    )

    safe_delete(stage_tif)

def run_steps_for_map(map_label, steps, step_offset=0):

    for i, fn in enumerate(steps, start=1):

        msg(f"{map_label}: step {i}/{len(steps)} ...")

        t0 = time.time()
        fn()
        dt = time.time() - t0

        msg(f"{map_label}: step {i}/{len(steps)} done ({dt:.1f} s)")
        arcpy.SetProgressorPosition(step_offset + i)


# -----------------------
# GENERIC VECTOR PROCESSOR
# -----------------------
def process_vector_job(job_name, in_fc, value_field, out_cog, stage_name, mask_sr, mask_raster, scratch_gdb, global_step):

    require_field(in_fc, value_field)

    safe_job = job_name.lower().replace(" ", "_")

    tmp_proj = os.path.join(scratch_gdb, f"{safe_job}_proj")
    tmp_ras = os.path.join(scratch_gdb, f"{safe_job}_ras")
    tmp_masked = os.path.join(scratch_gdb, f"{safe_job}_masked")

    for p in [tmp_proj, tmp_ras, tmp_masked]:
        safe_delete(p)

    def step1():
        arcpy.management.Project(in_fc, tmp_proj, mask_sr)

    def step2():
        arcpy.conversion.PolygonToRaster(
            in_features=tmp_proj,
            value_field=value_field,
            out_rasterdataset=tmp_ras,
            cell_assignment="MAXIMUM_AREA",
            priority_field="NONE",
            cellsize=arcpy.env.cellSize
        )

    def step3():
        from arcpy.sa import ExtractByMask
        ExtractByMask(tmp_ras, mask_raster).save(tmp_masked)

    def step4():
        to_cog(tmp_masked, out_cog, stage_name)

    msg(f"=== {job_name}: START ===")

    run_steps_for_map(job_name, [step1, step2, step3, step4], step_offset=global_step)

    msg(f"=== {job_name}: END ===")


# -----------------------
# MAIN
# -----------------------
def main():

    arcpy.env.overwriteOutput = True

    if not os.path.isdir(OUT_DIR):
        os.makedirs(OUT_DIR)

    arcpy.CheckOutExtension("Spatial")

    mask_desc = arcpy.Describe(MASK_RASTER)
    mask_sr = mask_desc.spatialReference
    mask_ras = arcpy.Raster(MASK_RASTER)

    # Environments aligned to mask
    arcpy.env.snapRaster = MASK_RASTER
    arcpy.env.cellSize = MASK_RASTER
    arcpy.env.extent = mask_ras.extent
    arcpy.env.outputCoordinateSystem = mask_sr
    arcpy.env.mask = MASK_RASTER

    # -----------------------
    # DEFINE JOBS
    # -----------------------
    all_jobs = [

        ("Marine", DO_MARINE, MARINE_SHP, MARINE_FIELD, OUT_MARINE_COG, "marine"),

        ("Landuse", DO_LANDUSE, LANDUSE_SHP, LANDUSE_FIELD, OUT_LANDUSE_COG, "landuse"),

        ("Geology_LogCa", DO_GEOLOGY_LOGCA, GEOLOGY_CA_FC, GEOLOGY_LOGCA_FIELD, OUT_GEOLOGY_LOGCA_COG, "geology_logca"),

        ("Geology_CaCoef", DO_GEOLOGY_CA_COE, GEOLOGY_CA_FC, GEOLOGY_CA_COE_FIELD, OUT_GEOLOGY_CA_COE_COG, "geology_cacoe"),

        ("Quaternary", DO_QUATERNARY, QUATERNARY_SHP, QUATERNARY_FIELD, OUT_QUATERNARY_COG, "quaternary")
    ]

    jobs = [j for j in all_jobs if j[1] and not should_skip(j[4])]

    msg("------------------------------------------------------------")
    msg(f"Planned outputs (maps to build): {len(jobs)}")

    if len(jobs) == 0:
        msg("Nothing to do (requested COGs already exist, or toggles off).")
        return

    msg("Will build: " + ", ".join([j[0] for j in jobs]))
    msg("------------------------------------------------------------")

    steps_per_map = 4
    total_steps = len(jobs) * steps_per_map

    arcpy.SetProgressor("step", "Rasterizing maps...", 0, total_steps, 1)

    scratch_gdb = arcpy.env.scratchGDB
    global_step = 0

    for name, _, fc, field, out, stage in jobs:

        process_vector_job(
            job_name=name,
            in_fc=fc,
            value_field=field,
            out_cog=out,
            stage_name=stage,
            mask_sr=mask_sr,
            mask_raster=MASK_RASTER,
            scratch_gdb=scratch_gdb,
            global_step=global_step
        )

        global_step += steps_per_map

    msg("------------------------------------------------------------")
    msg("Sanity check (dimensions):")

    msg(f"Mask rows/cols: {mask_ras.height} x {mask_ras.width}")

    for out in [
        OUT_MARINE_COG,
        OUT_LANDUSE_COG,
        OUT_GEOLOGY_LOGCA_COG,
        OUT_GEOLOGY_CA_COE_COG,
        OUT_QUATERNARY_COG
    ]:

        if os.path.exists(out):
            r = arcpy.Raster(out)
            msg(f"{os.path.basename(out)} rows/cols: {r.height} x {r.width}")

    arcpy.ResetProgressor()

    msg("Done.")
    msg("------------------------------------------------------------")


if __name__ == "__main__":
    main()
```

```python
###One hot encoding

import arcpy
import os
from arcpy.sa import ExtractByMask, Con, IsNull

# ============================================================
# USER INPUTS
# ============================================================

input_polygons = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Kalkinnhold_bergart_pol_clip.shp"
class_field = "kalkinnhol"

mask_raster = r"E:\Alpha_earth\qc\alphaearth_mosaic_epsg25833_band1_qc_cog.tif"

# ASCII-safe temp workspace
temp_root = r"C:\kalk_tmp"
temp_gdb = os.path.join(temp_root, "temp.gdb")

# Final output
output_folder = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Kalk_one_hot_encoding"

# ============================================================
# SETUP
# ============================================================

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")
arcpy.ClearWorkspaceCache_management()

os.makedirs(temp_root, exist_ok=True)
os.makedirs(output_folder, exist_ok=True)

if not arcpy.Exists(temp_gdb):
    arcpy.management.CreateFileGDB(temp_root, "temp.gdb")

# ============================================================
# ENVIRONMENT SETTINGS
# ============================================================

mask_desc = arcpy.Describe(mask_raster)

arcpy.env.snapRaster = mask_raster
arcpy.env.cellSize = mask_raster
arcpy.env.extent = mask_raster
arcpy.env.outputCoordinateSystem = mask_desc.spatialReference

# ============================================================
# CLASSES TO ENCODE (ONLY 1–5)
# ============================================================

target_classes = [1, 2, 3, 4, 5]

# ============================================================
# LOOP THROUGH CLASSES
# ============================================================

for cls in target_classes:
    print(f"\nProcessing class: {cls}")

    temp_layer = f"lyr_class_{cls}"
    temp_raster = os.path.join(temp_gdb, f"ras_class_{cls}")
    temp_masked = os.path.join(temp_gdb, f"ras_class_{cls}_mask")

    final_cog = os.path.join(output_folder, f"kalk_class_{cls}_cog.tif")

    where_clause = f"{arcpy.AddFieldDelimiters(input_polygons, class_field)} = {cls}"

    # --------------------------------------------------------
    # STEP 1 — SELECT CURRENT CLASS
    # --------------------------------------------------------

    arcpy.management.MakeFeatureLayer(input_polygons, temp_layer, where_clause)

    count = int(arcpy.management.GetCount(temp_layer)[0])
    if count == 0:
        print(f"  No polygons found for class {cls}, skipping.")
        arcpy.management.Delete(temp_layer)
        continue

    print(f"  Selected {count} polygons.")

    # --------------------------------------------------------
    # STEP 2 — RASTERIZE CURRENT CLASS
    # --------------------------------------------------------

    print("  Rasterizing selected polygons...")

    arcpy.conversion.PolygonToRaster(
        in_features=temp_layer,
        value_field=class_field,
        out_rasterdataset=temp_raster,
        cell_assignment="CELL_CENTER",
        priority_field=class_field,
        cellsize=mask_raster
    )

    # --------------------------------------------------------
    # STEP 3 — CONVERT TO BINARY
    # --------------------------------------------------------

    print("  Converting to binary raster...")

    binary_raster = Con(IsNull(temp_raster), 0, 1)

    # --------------------------------------------------------
    # STEP 4 — APPLY MASK
    # --------------------------------------------------------

    print("  Applying mask...")

    masked = ExtractByMask(binary_raster, mask_raster)
    masked.save(temp_masked)

    # --------------------------------------------------------
    # STEP 5 — EXPORT COG ONLY
    # --------------------------------------------------------

    print("  Exporting COG...")

    try:
        arcpy.management.CopyRaster(
            in_raster=temp_masked,
            out_rasterdataset=final_cog,
            format="COG",
            pixel_type="8_BIT_UNSIGNED",
            compression="LZW"
        )
        print("  COG created successfully.")
    except Exception as e:
        print(f"  COG export failed: {e}")

    arcpy.management.Delete(temp_layer)

# ============================================================
# CLEANUP
# ============================================================

arcpy.ClearWorkspaceCache_management()
arcpy.CheckInExtension("Spatial")

print("\nDONE")
```

```python
# ============================================================
# FAST ONE-HOT RASTERIZATION OF SELECTED/ALL CATEGORIES
# EXCEPT LAND USE / WITH TOGGLE FOR SELECTED CLASSES
# OPTIMIZED VERSION: NO REPEATED MASKING
# ============================================================

import arcpy
import os
import re
import csv
import time
import shutil
import gc

from datetime import datetime, timedelta
from arcpy.sa import ExtractByMask, Con, IsNull, SetNull, Raster


# ------------------------------------------------------------
# SPATIAL ANALYST
# ------------------------------------------------------------

arcpy.CheckOutExtension("Spatial")
arcpy.env.overwriteOutput = True


# ------------------------------------------------------------
# USER INPUTS
# ------------------------------------------------------------

mask_raster = r"E:\mask\Mask_land_Kalk_cog.tif"

input_shp = (
    r"N:\Prosjekter\411200_Kalk_i_grunnen\Berggrunn_N250_derivates"
    r"\BergFlate250_Hovedbergart_03sept26.shp"
)

category_field = "Forenklet1"

out_dir = r"G:\Geology_rasters_2026\Bedrock_edited"
os.makedirs(out_dir, exist_ok=True)

output_prefix = "Bedrock"


# ------------------------------------------------------------
# TOGGLE: SELECT ONLY SPECIFIC CLASSES?
# ------------------------------------------------------------

USE_SELECTED_CLASSES = False

# Write values exactly as they appear in category_field
SELECTED_CLASSES = [
    "kalkfyllitt, kalkglimmerskifer"
]


# ------------------------------------------------------------
# TEMPORARY WORKSPACE
# ------------------------------------------------------------

local_tmp_dir = r"C:\temp\onehot_Bedrock"
os.makedirs(local_tmp_dir, exist_ok=True)

tmp_gdb = os.path.join(
    local_tmp_dir,
    "temp_onehot.gdb"
)

print("\nPreparing temporary workspace:")
print(tmp_gdb)


# ------------------------------------------------------------
# RELEASE POSSIBLE ARCGIS LOCKS FROM PREVIOUS RUN
# ------------------------------------------------------------

arcpy.env.workspace = None
arcpy.env.scratchWorkspace = None

try:
    arcpy.ClearWorkspaceCache_management()
except Exception:
    pass

gc.collect()


# ------------------------------------------------------------
# DELETE OLD TEMPORARY GDB USING WINDOWS FILESYSTEM
#
# Do NOT use:
# arcpy.management.Delete(tmp_gdb)
#
# This avoids the GDB_Items SQL error encountered previously.
# ------------------------------------------------------------

if os.path.exists(tmp_gdb):

    print("\nOld temporary GDB found.")
    print("Deleting old temporary GDB...")

    try:
        shutil.rmtree(tmp_gdb)

        print("Old temporary GDB deleted successfully.")

    except Exception as e:

        raise RuntimeError(
            "\nCould not delete temporary geodatabase:\n"
            f"{tmp_gdb}\n\n"
            "The geodatabase may still be locked by ArcGIS Pro.\n"
            "Close any map/table using it or restart the Python kernel.\n\n"
            f"Original error:\n{e}"
        )


# ------------------------------------------------------------
# CREATE FRESH TEMPORARY GDB
# ------------------------------------------------------------

print("\nCreating fresh temporary geodatabase...")

arcpy.management.CreateFileGDB(
    local_tmp_dir,
    "temp_onehot.gdb"
)

print("Temporary GDB created:")
print(tmp_gdb)


# ------------------------------------------------------------
# ENVIRONMENT FROM MASK
# ------------------------------------------------------------

mask_desc = arcpy.Describe(mask_raster)

arcpy.env.workspace = tmp_gdb
arcpy.env.scratchWorkspace = tmp_gdb

arcpy.env.snapRaster = mask_raster
arcpy.env.cellSize = mask_raster
arcpy.env.extent = mask_raster
arcpy.env.outputCoordinateSystem = mask_desc.spatialReference
arcpy.env.mask = mask_raster


cell_size = float(
    arcpy.management.GetRasterProperties(
        mask_raster,
        "CELLSIZEX"
    ).getOutput(0)
)


print("\nMask raster:", mask_raster)
print("CRS:", mask_desc.spatialReference.name)
print("Cell size:", cell_size)

print(
    "Rows:",
    arcpy.management.GetRasterProperties(
        mask_raster,
        "ROWCOUNT"
    ).getOutput(0)
)

print(
    "Cols:",
    arcpy.management.GetRasterProperties(
        mask_raster,
        "COLUMNCOUNT"
    ).getOutput(0)
)


# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def safe_name(value, max_len=70):

    value = str(value).strip()

    replacements = {

        "æ": "ae",
        "ø": "o",
        "å": "a",

        "Æ": "Ae",
        "Ø": "O",
        "Å": "A",

        "é": "e",
        "è": "e",
        "ê": "e",

        "É": "E",
        "È": "E",
        "Ê": "E",

        "'": "",
        '"': ""
    }

    for old, new in replacements.items():
        value = value.replace(old, new)

    value = re.sub(
        r"[^A-Za-z0-9_]+",
        "_",
        value
    )

    value = re.sub(
        r"_+",
        "_",
        value
    )

    value = value.strip("_")

    return value[:max_len] if value else "unknown"



def format_seconds(seconds):

    seconds = int(seconds)

    if seconds < 60:
        return f"{seconds} sec"

    minutes = seconds / 60

    if minutes < 60:
        return f"{minutes:.1f} min"

    return f"{minutes / 60:.2f} hr"



def progress_bar(current, total, width=30):

    fraction = current / total if total else 1

    filled = int(width * fraction)

    bar = (
        "#" * filled
        +
        "-" * (width - filled)
    )

    return (
        f"[{bar}] "
        f"{100 * fraction:5.1f}%"
    )



def sql_value(v, field_type):

    if field_type in [
        "Integer",
        "SmallInteger",
        "Double",
        "Single"
    ]:

        return str(v)

    else:

        return (
            "'"
            +
            str(v).replace(
                "'",
                "''"
            )
            +
            "'"
        )


# ------------------------------------------------------------
# CHECK INPUT
# ------------------------------------------------------------

if not arcpy.Exists(input_shp):

    raise FileNotFoundError(
        f"Input shapefile not found:\n{input_shp}"
    )


fields = arcpy.ListFields(input_shp)

field_names = [
    f.name
    for f in fields
]


if category_field not in field_names:

    print("\nAvailable fields:")

    for f in field_names:
        print(" -", f)

    raise ValueError(
        f"\nField '{category_field}' not found in:\n"
        f"{input_shp}"
    )


field_type = [
    f.type
    for f in fields
    if f.name == category_field
][0]


# ------------------------------------------------------------
# PROJECT TO MASK CRS IF NEEDED
# ------------------------------------------------------------

in_desc = arcpy.Describe(input_shp)

projected_fc = os.path.join(
    tmp_gdb,
    f"{output_prefix}_projected"
)


if (
    in_desc.spatialReference.name
    !=
    mask_desc.spatialReference.name
):

    print(
        "\nProjecting polygon layer "
        "to mask CRS..."
    )

    arcpy.management.Project(
        input_shp,
        projected_fc,
        mask_desc.spatialReference
    )

    work_fc = projected_fc


else:

    print(
        "\nInput already has same CRS as mask."
    )

    work_fc = input_shp


# ------------------------------------------------------------
# COPY FEATURES LOCALLY
# ------------------------------------------------------------

local_fc = os.path.join(
    tmp_gdb,
    f"{output_prefix}_local"
)


print(
    "\nCopying polygons to local "
    "temporary geodatabase..."
)


arcpy.management.CopyFeatures(
    work_fc,
    local_fc
)


# ------------------------------------------------------------
# OPTIONAL: SELECT ONLY SPECIFIC CLASSES
# ------------------------------------------------------------

if USE_SELECTED_CLASSES:

    selected_fc = os.path.join(
        tmp_gdb,
        f"{output_prefix}_selected"
    )


    values_sql = ",".join(
        sql_value(
            v,
            field_type
        )
        for v in SELECTED_CLASSES
    )


    field_sql = arcpy.AddFieldDelimiters(
        local_fc,
        category_field
    )


    where_clause = (
        f"{field_sql} "
        f"IN ({values_sql})"
    )


    print(
        "\nSelecting only requested classes:"
    )

    print(where_clause)


    arcpy.management.MakeFeatureLayer(
        local_fc,
        "selected_layer",
        where_clause
    )


    arcpy.management.CopyFeatures(
        "selected_layer",
        selected_fc
    )


    selected_count = int(
        arcpy.management.GetCount(
            selected_fc
        ).getOutput(0)
    )


    print(
        "Selected polygons:",
        selected_count
    )


    if selected_count == 0:

        raise ValueError(
            "No polygons selected. "
            "Check SELECTED_CLASSES and "
            "category_field values."
        )


    local_fc = selected_fc


# ------------------------------------------------------------
# UNIQUE CATEGORIES
# ------------------------------------------------------------

categories = sorted({

    row[0]

    for row
    in arcpy.da.SearchCursor(
        local_fc,
        [category_field]
    )

    if row[0] not in [
        None,
        "",
        " "
    ]
})


cat_to_id = {

    cat: i + 1

    for i, cat
    in enumerate(categories)
}


print(
    f"\nNumber of categories to rasterize "
    f"in {category_field}: "
    f"{len(categories)}"
)


for cat, class_id in cat_to_id.items():

    print(
        f" {class_id:02d} = {cat}"
    )


# ------------------------------------------------------------
# LOOKUP CSV
# ------------------------------------------------------------

lookup_csv = os.path.join(
    out_dir,
    f"{output_prefix}_category_lookup.csv"
)


with open(
    lookup_csv,
    "w",
    newline="",
    encoding="utf-8"
) as f:

    writer = csv.writer(f)

    writer.writerow([
        "class_id",
        "category",
        "output_raster"
    ])


    for cat, class_id in cat_to_id.items():

        cat_safe = safe_name(cat)

        out_name = (
            f"{output_prefix}_"
            f"{class_id:02d}_"
            f"{cat_safe}.tif"
        )

        writer.writerow([
            class_id,
            cat,
            out_name
        ])


print(
    "\nSaved category lookup table:"
)

print(lookup_csv)


# ------------------------------------------------------------
# ADD CLASS ID FIELD
# ------------------------------------------------------------

class_field = "class_id"


if class_field not in [
    f.name
    for f in arcpy.ListFields(local_fc)
]:

    arcpy.management.AddField(
        local_fc,
        class_field,
        "SHORT"
    )


print(
    "\nAssigning integer class IDs..."
)


with arcpy.da.UpdateCursor(
    local_fc,
    [
        category_field,
        class_field
    ]
) as cursor:

    for row in cursor:

        row[1] = cat_to_id.get(
            row[0],
            0
        )

        cursor.updateRow(row)


# ------------------------------------------------------------
# RASTERIZE POLYGONS ONCE
# ------------------------------------------------------------

class_raster = os.path.join(
    tmp_gdb,
    f"{output_prefix}_class_raster"
)


print(
    f"\nRasterizing {output_prefix} polygons "
    "once using class_id..."
)


rasterize_start = time.time()


arcpy.conversion.PolygonToRaster(

    in_features=local_fc,

    value_field=class_field,

    out_rasterdataset=class_raster,

    cell_assignment="MAXIMUM_AREA",

    priority_field="NONE",

    cellsize=cell_size
)


print(
    "Rasterization completed in "
    f"{format_seconds(time.time() - rasterize_start)}"
)


# ------------------------------------------------------------
# MASK CLASS RASTER ONCE
# AND FILL NON-SELECTED LAND CELLS WITH 0
# ------------------------------------------------------------

print(
    "\nMasking integer class raster once "
    "and filling non-selected land cells with 0..."
)


mask_start = time.time()


mask_r = Raster(mask_raster)


class_raster_masked = ExtractByMask(
    class_raster,
    mask_raster
)


class_raster_full = SetNull(

    IsNull(mask_r),

    Con(
        IsNull(class_raster_masked),
        0,
        class_raster_masked
    )
)


print(
    "Masking completed in "
    f"{format_seconds(time.time() - mask_start)}"
)


# ------------------------------------------------------------
# SAVE INTEGER CLASS COG
# ------------------------------------------------------------

class_raster_tif = os.path.join(
    out_dir,
    f"{output_prefix}_class_id.tif"
)


if arcpy.Exists(class_raster_tif):

    arcpy.management.Delete(
        class_raster_tif
    )


print(
    "\nSaving integer class raster as COG:"
)

print(class_raster_tif)


save_start = time.time()


arcpy.management.CopyRaster(

    in_raster=class_raster_full,

    out_rasterdataset=class_raster_tif,

    pixel_type="16_BIT_UNSIGNED",

    nodata_value="65535",

    format="COG"
)


print(
    "Integer class raster saved in "
    f"{format_seconds(time.time() - save_start)}"
)


# ------------------------------------------------------------
# CREATE BINARY COG RASTERS
# ------------------------------------------------------------

print(
    "\nCreating binary one-hot rasters..."
)

print(
    "Binary rasters: "
    "1 = selected class, "
    "0 = other landmask pixels, "
    "NoData = outside landmask."
)


total_maps = len(cat_to_id)

overall_start = time.time()

map_times = []


for idx, (cat, class_id) in enumerate(
    cat_to_id.items(),
    start=1
):

    map_start = time.time()

    cat_safe = safe_name(cat)


    out_cog = os.path.join(

        out_dir,

        f"{output_prefix}_"
        f"{class_id:02d}_"
        f"{cat_safe}.tif"
    )


    print(
        "\n" + "=" * 80
    )

    print(
        f"[{idx}/{total_maps}] {cat}"
    )

    print(
        progress_bar(
            idx - 1,
            total_maps
        )
    )

    print(
        "=" * 80
    )


    # --------------------------------------------------------
    # REMOVE EXISTING OUTPUT
    # --------------------------------------------------------

    if arcpy.Exists(out_cog):

        arcpy.management.Delete(
            out_cog
        )


    # --------------------------------------------------------
    # BINARY CLASS MAP
    #
    # 1 = this class
    # 0 = all other valid land pixels
    # NoData = outside mask
    # --------------------------------------------------------

    binary = SetNull(

        IsNull(mask_r),

        Con(
            class_raster_full == class_id,
            1,
            0
        )
    )


    # --------------------------------------------------------
    # SAVE AS COG
    # --------------------------------------------------------

    arcpy.management.CopyRaster(

        in_raster=binary,

        out_rasterdataset=out_cog,

        pixel_type="8_BIT_UNSIGNED",

        nodata_value="255",

        format="COG"
    )


    # --------------------------------------------------------
    # PROGRESS / ETA
    # --------------------------------------------------------

    elapsed_map = (
        time.time()
        -
        map_start
    )


    map_times.append(
        elapsed_map
    )


    completed = idx

    remaining = (
        total_maps
        -
        completed
    )


    avg_time = (
        sum(map_times)
        /
        len(map_times)
    )


    eta_sec = (
        avg_time
        *
        remaining
    )


    finish_time = (
        datetime.now()
        +
        timedelta(
            seconds=eta_sec
        )
    )


    print(
        "\nSaved:",
        os.path.basename(out_cog)
    )


    print(
        progress_bar(
            completed,
            total_maps
        )
    )


    print(
        f"Map time:         "
        f"{format_seconds(elapsed_map)}"
    )

    print(
        f"Average/map:      "
        f"{format_seconds(avg_time)}"
    )

    print(
        f"ETA remaining:    "
        f"{format_seconds(eta_sec)}"
    )

    print(
        "Estimated finish: "
        f"{finish_time.strftime('%Y-%m-%d %H:%M:%S')}"
    )


# ------------------------------------------------------------
# FINISHED
# ------------------------------------------------------------

total_elapsed = (
    time.time()
    -
    overall_start
)


print(
    "\n" + "=" * 80
)

print("Done.")

print(
    "Total binary raster processing time:",
    format_seconds(total_elapsed)
)

print(
    "Output folder:",
    out_dir
)

print(
    "Lookup CSV:",
    lookup_csv
)

print(
    "=" * 80
)
```

```python
##########################################################
##########Create a raster map with NO one-hot encoding_EDITED_FOR_TEXT AND NUMERICAL ENTRIE: Polygon to Raster masked
###########################################################

import arcpy
import os
import csv
from arcpy.sa import ExtractByMask

# ============================================================
# USER INPUTS
# ============================================================

input_polygons = r"N:\Prosjekter\411200_Kalk_i_grunnen\Berggrunn_N250_derivates\BergFlate250_Hovedbergart.shp"

class_field = "Forenk_Ca"

mask_raster = r"E:\Alpha_earth\qc\alphaearth_mosaic_epsg25833_band1_qc_cog.tif"

temp_root = r"C:\Bergart_tmp"
temp_gdb = os.path.join(temp_root, "temp.gdb")

output_folder = r"G:\Geology_rasters_2026\Bedrock"

output_raster = os.path.join(
    output_folder,
    "Bergart_250_forenklet.tif"
)

lookup_csv = os.path.join(
    output_folder,
    "Bergart_250_forenklet_lookup.csv"
)

# ============================================================
# SETUP
# ============================================================

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

os.makedirs(temp_root, exist_ok=True)
os.makedirs(output_folder, exist_ok=True)

if not arcpy.Exists(temp_gdb):
    arcpy.management.CreateFileGDB(temp_root, "temp.gdb")

temp_raster = os.path.join(temp_gdb, "bergart_raster")
temp_masked = os.path.join(temp_gdb, "bergart_raster_masked")

# ============================================================
# ENVIRONMENT SETTINGS
# ============================================================

mask_desc = arcpy.Describe(mask_raster)

arcpy.env.snapRaster = mask_raster
arcpy.env.cellSize = mask_raster
arcpy.env.extent = mask_raster
arcpy.env.outputCoordinateSystem = mask_desc.spatialReference

# ============================================================
# DETERMINE FIELD TYPE
# ============================================================

field_obj = [f for f in arcpy.ListFields(input_polygons)
             if f.name == class_field][0]

field_type = field_obj.type

print(f"Field: {class_field}")
print(f"Field type: {field_type}")

# ============================================================
# CREATE NUMERIC FIELD IF NEEDED
# ============================================================

raster_field = class_field

numeric_types = [
    "SmallInteger",
    "Integer",
    "Single",
    "Double",
    "BigInteger"
]

if field_type not in numeric_types:

    print("Text field detected. Creating integer codes...")

    raster_field = "RASTER_ID"

    existing_fields = [f.name for f in arcpy.ListFields(input_polygons)]

    if raster_field not in existing_fields:
        arcpy.management.AddField(
            input_polygons,
            raster_field,
            "LONG"
        )

    unique_values = set()

    with arcpy.da.SearchCursor(
        input_polygons,
        [class_field]
    ) as cursor:

        for row in cursor:
            if row[0] is not None:
                unique_values.add(str(row[0]))

    unique_values = sorted(list(unique_values))

    value_to_code = {
        value: idx + 1
        for idx, value in enumerate(unique_values)
    }

    with arcpy.da.UpdateCursor(
        input_polygons,
        [class_field, raster_field]
    ) as cursor:

        for row in cursor:
            row[1] = value_to_code.get(str(row[0]), 0)
            cursor.updateRow(row)

    print("Writing lookup table...")

    with open(
        lookup_csv,
        "w",
        newline="",
        encoding="utf-8"
    ) as csvfile:

        writer = csv.writer(csvfile)

        writer.writerow([
            "Raster_Code",
            "Original_Label"
        ])

        for label, code in value_to_code.items():
            writer.writerow([code, label])

    print("Lookup table saved:")
    print(lookup_csv)

else:

    print("Numeric field detected. No recoding required.")

# ============================================================
# RASTERIZE
# ============================================================

print("Rasterizing polygons...")

arcpy.conversion.PolygonToRaster(
    in_features=input_polygons,
    value_field=raster_field,
    out_rasterdataset=temp_raster,
    cell_assignment="CELL_CENTER",
    cellsize=mask_raster
)

# ============================================================
# APPLY MASK
# ============================================================

print("Applying mask...")

masked = ExtractByMask(temp_raster, mask_raster)
masked.save(temp_masked)

# ============================================================
# EXPORT TIFF
# ============================================================

print("Exporting TIFF...")

arcpy.management.CopyRaster(
    in_raster=temp_masked,
    out_rasterdataset=output_raster,
    format="TIFF"
)

# ============================================================
# CLEANUP
# ============================================================

arcpy.ClearWorkspaceCache_management()
arcpy.CheckInExtension("Spatial")

print("\nDONE")
print("Raster:")
print(output_raster)

if os.path.exists(lookup_csv):
    print("\nLookup table:")
    print(lookup_csv)
```

```python
# ============================================================
# ONE-HOT RASTERIZATION OF SELECTED LANDUSE CATEGORIES
# Optimized: rasterize once, then create binary COGs
# ============================================================

import arcpy
import os
import re
import csv
import time
from datetime import datetime, timedelta
from arcpy.sa import ExtractByMask, Con, IsNull, SetNull, Raster

arcpy.CheckOutExtension("Spatial")
arcpy.env.overwriteOutput = True

# ------------------------------------------------------------
# USER INPUTS
# ------------------------------------------------------------

mask_raster = r"E:\mask\Mask_land_Kalk_cog.tif"

input_shp = (
    r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Landuse\Norway_corine"
    r"\Norway_landuse.shp"
)

category_field = "Code_18"

out_dir = r"G:\Geology_rasters_2026\Landuse_selected"
os.makedirs(out_dir, exist_ok=True)

local_tmp_dir = r"C:\temp\onehot_landuse"
os.makedirs(local_tmp_dir, exist_ok=True)

tmp_gdb = os.path.join(local_tmp_dir, "temp_onehot_landuse.gdb")

if arcpy.Exists(tmp_gdb):
    arcpy.management.Delete(tmp_gdb)

arcpy.management.CreateFileGDB(local_tmp_dir, "temp_onehot_landuse.gdb")

# ------------------------------------------------------------
# SELECTED LANDUSE CLASSES FOR Ca-AFFINITY NATURTYPE MODELLING
# Code_18 values from your CLC legend
# ------------------------------------------------------------

selected_classes = {
    #111: "Continuous urban fabric",
    #112: "Discontinuous Urban Fabric",
    121: "Industrial_or_commercial_units",
    122: "Road_and_rail_networks_and_associated_land",
    123: "Port_areas",
    124: "Airports",
    131: "Mineral_extraction_sites",
    132: "Dump_sites",
    133: "Construction_sites",
    141: "Green_urban_areas",
    142: "Sport and leisure facilities",
    #211: "Non-irrigated arable land",
    #231: "Pastures",
    #242: "Agricultural_area",
    #243: "Agriculture_with_significant_natural_vegetation",
    #244: "Agro_forestry_areas",
    #311: "Broad_leaved_forest",
    #312: "Coniferous_forest",
    #313: "Mixed_forest",
    #321: "Natural_grasslands",
    #322: "Moors_and_heathland",
    #324: "Transitional_woodland_shrub",
    #331: "Beaches_dunes_sands",
    #332: "Bare_rocks",
    #333: "Sparsely_vegetated_areas",
    334: "Burnt_areas",
    335: "Glaciers_and_perpetual_snow",
    #411: "Inland_marshes",
    #412: "Peat_bogs",
    #421: "Salt_marshes",
    #423: "Intertidal_flats",
    511: "Water_courses",
    512: "Water_bodies",
    522: "Estuaries",
    523: "Sea_and_ocean",
}
# ------------------------------------------------------------
# PRINT REQUESTED CLASSES IMMEDIATELY
# ------------------------------------------------------------

print("\nRequested selected landuse classes:")
print("-" * 80)
for code, label in selected_classes.items():
    print(f"Code_18={code} | {label}")
print("-" * 80)
print(f"Total requested classes: {len(selected_classes)}")

# ------------------------------------------------------------
# ENVIRONMENT FROM MASK
# ------------------------------------------------------------

mask_desc = arcpy.Describe(mask_raster)

arcpy.env.workspace = tmp_gdb
arcpy.env.scratchWorkspace = tmp_gdb
arcpy.env.snapRaster = mask_raster
arcpy.env.cellSize = mask_raster
arcpy.env.extent = mask_raster
arcpy.env.outputCoordinateSystem = mask_desc.spatialReference
arcpy.env.mask = mask_raster

cell_size = float(
    arcpy.management.GetRasterProperties(mask_raster, "CELLSIZEX").getOutput(0)
)

print("Mask raster:", mask_raster)
print("CRS:", mask_desc.spatialReference.name)
print("Cell size:", cell_size)
print("Rows:", arcpy.management.GetRasterProperties(mask_raster, "ROWCOUNT").getOutput(0))
print("Cols:", arcpy.management.GetRasterProperties(mask_raster, "COLUMNCOUNT").getOutput(0))

# ------------------------------------------------------------
# HELPERS
# ------------------------------------------------------------

def safe_name(value, max_len=70):
    value = str(value).strip()
    value = re.sub(r"[^A-Za-z0-9_]+", "_", value)
    value = re.sub(r"_+", "_", value)
    value = value.strip("_")
    return value[:max_len]


def format_seconds(seconds):
    seconds = int(seconds)
    if seconds < 60:
        return f"{seconds} sec"
    minutes = seconds / 60
    if minutes < 60:
        return f"{minutes:.1f} min"
    return f"{minutes / 60:.2f} hr"


def progress_bar(current, total, width=30):
    fraction = current / total
    filled = int(width * fraction)
    bar = "#" * filled + "-" * (width - filled)
    return f"[{bar}] {100 * fraction:5.1f}%"


# ------------------------------------------------------------
# CHECK INPUT
# ------------------------------------------------------------

if not arcpy.Exists(input_shp):
    raise FileNotFoundError(f"Input shapefile not found:\n{input_shp}")

fields = [f.name for f in arcpy.ListFields(input_shp)]

if category_field not in fields:
    print("\nAvailable fields:")
    for f in fields:
        print(" -", f)
    raise ValueError(f"Field '{category_field}' not found.")

# ------------------------------------------------------------
# PROJECT IF NEEDED
# ------------------------------------------------------------

in_desc = arcpy.Describe(input_shp)
projected_fc = os.path.join(tmp_gdb, "landuse_projected")

if in_desc.spatialReference.name != mask_desc.spatialReference.name:
    print("\nProjecting landuse polygons to mask CRS...")
    arcpy.management.Project(input_shp, projected_fc, mask_desc.spatialReference)
    work_fc = projected_fc
else:
    print("\nInput already has same CRS as mask.")
    work_fc = input_shp

# ------------------------------------------------------------
# COPY LOCALLY
# ------------------------------------------------------------

local_fc = os.path.join(tmp_gdb, "landuse_local")

print("\nCopying polygons to local temporary geodatabase...")
arcpy.management.CopyFeatures(work_fc, local_fc)

# ------------------------------------------------------------
# KEEP ONLY SELECTED CATEGORIES
# ------------------------------------------------------------

selected_fc = os.path.join(tmp_gdb, "landuse_selected")

field_type = [f.type for f in arcpy.ListFields(local_fc) if f.name == category_field][0]

if field_type in ["Integer", "SmallInteger", "Double", "Single"]:
    selected_values_sql = ",".join(str(v) for v in selected_classes.keys())
    where_clause = f"{arcpy.AddFieldDelimiters(local_fc, category_field)} IN ({selected_values_sql})"
else:
    selected_values_sql = ",".join(f"'{v}'" for v in selected_classes.keys())
    where_clause = f"{arcpy.AddFieldDelimiters(local_fc, category_field)} IN ({selected_values_sql})"

print("\nSelecting useful landuse classes only...")
print(where_clause)

arcpy.management.MakeFeatureLayer(local_fc, "landuse_lyr", where_clause)
arcpy.management.CopyFeatures("landuse_lyr", selected_fc)

count_selected = int(arcpy.management.GetCount(selected_fc).getOutput(0))
print("Selected polygons:", count_selected)

if count_selected == 0:
    raise ValueError("No polygons selected. Check whether Code_18 is numeric or uses different codes.")

# ------------------------------------------------------------
# BUILD CATEGORY IDS
# ------------------------------------------------------------

found_codes = sorted({
    int(row[0])
    for row in arcpy.da.SearchCursor(selected_fc, [category_field])
    if row[0] is not None
})

code_to_class_id = {code: i + 1 for i, code in enumerate(found_codes)}

print("\nSelected categories found:")
for code in found_codes:
    print(f" {code_to_class_id[code]:02d} | Code_18={code} | {selected_classes.get(code, 'Unknown')}")

# ------------------------------------------------------------
# LOOKUP CSV
# ------------------------------------------------------------

lookup_csv = os.path.join(out_dir, "landuse_selected_category_lookup.csv")

with open(lookup_csv, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["class_id", "Code_18", "landuse_class", "output_raster"])

    for code in found_codes:
        class_id = code_to_class_id[code]
        label = selected_classes.get(code, "Unknown")
        out_name = f"landuse_{class_id:02d}_Code18_{code}_{safe_name(label)}.tif"
        writer.writerow([class_id, code, label, out_name])

print("\nSaved lookup table:")
print(lookup_csv)

# ------------------------------------------------------------
# ADD CLASS ID FIELD
# ------------------------------------------------------------

class_field = "class_id"

if class_field not in [f.name for f in arcpy.ListFields(selected_fc)]:
    arcpy.management.AddField(selected_fc, class_field, "SHORT")

print("\nAssigning class IDs...")

with arcpy.da.UpdateCursor(selected_fc, [category_field, class_field]) as cursor:
    for row in cursor:
        code = int(row[0])
        row[1] = code_to_class_id.get(code, 0)
        cursor.updateRow(row)

# ------------------------------------------------------------
# RASTERIZE ONCE
# ------------------------------------------------------------

class_raster = os.path.join(tmp_gdb, "landuse_selected_class_raster")

print("\nRasterizing selected landuse polygons once...")
rasterize_start = time.time()

arcpy.conversion.PolygonToRaster(
    in_features=selected_fc,
    value_field=class_field,
    out_rasterdataset=class_raster,
    cell_assignment="MAXIMUM_AREA",
    priority_field="NONE",
    cellsize=cell_size
)

print(f"Rasterization completed in {format_seconds(time.time() - rasterize_start)}")

# ------------------------------------------------------------
# MASK ONCE AND FILL NON-SELECTED LAND CELLS WITH 0
# ------------------------------------------------------------

print("\nMasking integer class raster once and filling non-selected land cells with 0...")
mask_start = time.time()

mask_r = Raster(mask_raster)

class_raster_masked = ExtractByMask(class_raster, mask_raster)

class_raster_full = SetNull(
    IsNull(mask_r),
    Con(IsNull(class_raster_masked), 0, class_raster_masked)
)

print(f"Masking completed in {format_seconds(time.time() - mask_start)}")

# ------------------------------------------------------------
# SAVE INTEGER CLASS COG
# ------------------------------------------------------------

class_raster_tif = os.path.join(out_dir, "landuse_selected_class_id.tif")

if arcpy.Exists(class_raster_tif):
    arcpy.management.Delete(class_raster_tif)

print("\nSaving integer class raster as COG:")
print(class_raster_tif)

save_start = time.time()

arcpy.management.CopyRaster(
    in_raster=class_raster_full,
    out_rasterdataset=class_raster_tif,
    pixel_type="16_BIT_UNSIGNED",
    nodata_value="65535",
    format="COG"
)

print(f"Integer class raster saved in {format_seconds(time.time() - save_start)}")

# ------------------------------------------------------------
# CREATE BINARY COG RASTERS
# ------------------------------------------------------------

print("\nCreating binary one-hot rasters...")
print("Repeated ExtractByMask is skipped because class raster is already masked.")
print("Binary rasters: 1 = selected class, 0 = other landmask pixels, NoData = outside landmask.")

total_maps = len(found_codes)
overall_start = time.time()
map_times = []

for idx, code in enumerate(found_codes, start=1):

    map_start = time.time()

    class_id = code_to_class_id[code]
    label = selected_classes.get(code, "Unknown")

    out_cog = os.path.join(
        out_dir,
        f"landuse_{class_id:02d}_Code18_{code}_{safe_name(label)}.tif"
    )

    print("\n" + "=" * 80)
    print(f"[{idx}/{total_maps}] Code_18={code} | {label}")
    print(progress_bar(idx - 1, total_maps))
    print("=" * 80)

    if arcpy.Exists(out_cog):
        arcpy.management.Delete(out_cog)

    binary = SetNull(
        IsNull(mask_r),
        Con(class_raster_full == class_id, 1, 0)
    )

    arcpy.management.CopyRaster(
        in_raster=binary,
        out_rasterdataset=out_cog,
        pixel_type="8_BIT_UNSIGNED",
        nodata_value="255",
        format="COG"
    )

    elapsed_map = time.time() - map_start
    map_times.append(elapsed_map)

    completed = idx
    remaining = total_maps - completed
    avg_time = sum(map_times) / len(map_times)
    eta_sec = avg_time * remaining
    finish_time = datetime.now() + timedelta(seconds=eta_sec)

    print("\nSaved:", os.path.basename(out_cog))
    print(progress_bar(completed, total_maps))
    print(f"Map time:         {format_seconds(elapsed_map)}")
    print(f"Average/map:      {format_seconds(avg_time)}")
    print(f"ETA remaining:    {format_seconds(eta_sec)}")
    print(f"Estimated finish: {finish_time.strftime('%Y-%m-%d %H:%M:%S')}")

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

total_elapsed = time.time() - overall_start
avg_elapsed = total_elapsed / total_maps if total_maps > 0 else 0

print("\n" + "=" * 80)
print("PROCESS COMPLETED")
print("=" * 80)
print(f"Total binary rasters: {total_maps}")
print(f"Total binary time:    {format_seconds(total_elapsed)}")
print(f"Output directory:     {out_dir}")
print(f"Lookup table:         {lookup_csv}")
print("=" * 80)
```
