```python
###Rasterize category maps

# -*- coding: utf-8 -*-
import os
import time
import arcpy

# -----------------------
# TOGGLES
# -----------------------
DO_MARINE = False
DO_LANDUSE = False

DO_GEOLOGY_LOGCA = True
DO_GEOLOGY_CA_COE = True

DO_QUATERNARY = True

OVERWRITE_EXISTING = False  # False = skip if output exists

# -----------------------
# INPUTS
# -----------------------
MARINE_SHP = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Marine_limit\Marine_limit_to_raster.shp"

LANDUSE_SHP = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Landuse\Norway_corine\Norway_landuse.shp"

# IMPORTANT: shapefile must include .shp
GEOLOGY_CA_FC = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Berggrunns_Kalk_feb2026_clip.shp"

QUATERNARY_SHP = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\Quaternary\LosmassteFlate_Lokal-Regional_forenklet_feb2026.shp"

MASK_RASTER = r"E:\Alpha_earth\qc\alphaearth_mosaic_epsg25833_band1_qc_cog.tif"

OUT_DIR = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Category_maps"

# -----------------------
# FIELDS
# -----------------------
MARINE_FIELD = "ANNETKVTEM"
LANDUSE_FIELD = "Code_18"

GEOLOGY_LOGCA_FIELD = "LogCa_ICP_"
GEOLOGY_CA_COE_FIELD = "Ca_ICP_Coe"

QUATERNARY_FIELD = "ForenkletK"

# -----------------------
# OUTPUTS
# -----------------------
OUT_MARINE_COG = os.path.join(OUT_DIR, "marine_limit_ANNETKVTEM_cog.tif")
OUT_LANDUSE_COG = os.path.join(OUT_DIR, "landuse_Code_18_cog.tif")

OUT_GEOLOGY_LOGCA_COG = os.path.join(OUT_DIR, "geology_LogCa_ICP_cog.tif")
OUT_GEOLOGY_CA_COE_COG = os.path.join(OUT_DIR, "geology_Ca_ICP_Coe_cog.tif")

OUT_QUATERNARY_COG = os.path.join(OUT_DIR, "quaternary_ForenkletK_cog.tif")


# -----------------------
# HELPERS
# -----------------------
def ts():
    return time.strftime("%Y-%m-%d %H:%M:%S")

def msg(s):
    arcpy.AddMessage(f"[{ts()}] {s}")
    print(f"[{ts()}] {s}")

def list_fields(fc):
    return [f.name for f in arcpy.ListFields(fc)]

def require_field(fc, field_name):
    if field_name not in list_fields(fc):
        msg(f"Fields in {os.path.basename(fc)}:")
        for fn in list_fields(fc):
            msg(f"  - {fn}")
        raise ValueError(f"Field '{field_name}' not found in: {fc}")

def safe_delete(path):
    try:
        if arcpy.Exists(path):
            arcpy.management.Delete(path)
        elif os.path.exists(path):
            os.remove(path)
    except Exception:
        pass

def should_skip(out_path):
    return (not OVERWRITE_EXISTING) and os.path.exists(out_path)

def stage_geotiff(in_raster, stage_tif):
    safe_delete(stage_tif)
    arcpy.management.CopyRaster(
        in_raster=in_raster,
        out_rasterdataset=stage_tif,
        format="TIFF"
    )

def to_cog(in_raster, out_cog, stage_name):

    if should_skip(out_cog):
        msg(f"Skipping (already exists): {out_cog}")
        return

    if not os.path.isdir(os.path.dirname(out_cog)):
        os.makedirs(os.path.dirname(out_cog))

    if OVERWRITE_EXISTING:
        safe_delete(out_cog)

    scratch_folder = arcpy.env.scratchFolder or os.environ.get("TEMP", r"C:\Temp")
    stage_tif = os.path.join(scratch_folder, f"{stage_name}_stage.tif")

    msg(f"Staging GeoTIFF: {stage_tif}")
    stage_geotiff(in_raster, stage_tif)

    msg(f"Writing COG: {out_cog}")
    safe_delete(out_cog)

    arcpy.management.CopyRaster(
        in_raster=stage_tif,
        out_rasterdataset=out_cog,
        format="COG"
    )

    safe_delete(stage_tif)

def run_steps_for_map(map_label, steps, step_offset=0):

    for i, fn in enumerate(steps, start=1):

        msg(f"{map_label}: step {i}/{len(steps)} ...")

        t0 = time.time()
        fn()
        dt = time.time() - t0

        msg(f"{map_label}: step {i}/{len(steps)} done ({dt:.1f} s)")
        arcpy.SetProgressorPosition(step_offset + i)


# -----------------------
# GENERIC VECTOR PROCESSOR
# -----------------------
def process_vector_job(job_name, in_fc, value_field, out_cog, stage_name, mask_sr, mask_raster, scratch_gdb, global_step):

    require_field(in_fc, value_field)

    safe_job = job_name.lower().replace(" ", "_")

    tmp_proj = os.path.join(scratch_gdb, f"{safe_job}_proj")
    tmp_ras = os.path.join(scratch_gdb, f"{safe_job}_ras")
    tmp_masked = os.path.join(scratch_gdb, f"{safe_job}_masked")

    for p in [tmp_proj, tmp_ras, tmp_masked]:
        safe_delete(p)

    def step1():
        arcpy.management.Project(in_fc, tmp_proj, mask_sr)

    def step2():
        arcpy.conversion.PolygonToRaster(
            in_features=tmp_proj,
            value_field=value_field,
            out_rasterdataset=tmp_ras,
            cell_assignment="MAXIMUM_AREA",
            priority_field="NONE",
            cellsize=arcpy.env.cellSize
        )

    def step3():
        from arcpy.sa import ExtractByMask
        ExtractByMask(tmp_ras, mask_raster).save(tmp_masked)

    def step4():
        to_cog(tmp_masked, out_cog, stage_name)

    msg(f"=== {job_name}: START ===")

    run_steps_for_map(job_name, [step1, step2, step3, step4], step_offset=global_step)

    msg(f"=== {job_name}: END ===")


# -----------------------
# MAIN
# -----------------------
def main():

    arcpy.env.overwriteOutput = True

    if not os.path.isdir(OUT_DIR):
        os.makedirs(OUT_DIR)

    arcpy.CheckOutExtension("Spatial")

    mask_desc = arcpy.Describe(MASK_RASTER)
    mask_sr = mask_desc.spatialReference
    mask_ras = arcpy.Raster(MASK_RASTER)

    # Environments aligned to mask
    arcpy.env.snapRaster = MASK_RASTER
    arcpy.env.cellSize = MASK_RASTER
    arcpy.env.extent = mask_ras.extent
    arcpy.env.outputCoordinateSystem = mask_sr
    arcpy.env.mask = MASK_RASTER

    # -----------------------
    # DEFINE JOBS
    # -----------------------
    all_jobs = [

        ("Marine", DO_MARINE, MARINE_SHP, MARINE_FIELD, OUT_MARINE_COG, "marine"),

        ("Landuse", DO_LANDUSE, LANDUSE_SHP, LANDUSE_FIELD, OUT_LANDUSE_COG, "landuse"),

        ("Geology_LogCa", DO_GEOLOGY_LOGCA, GEOLOGY_CA_FC, GEOLOGY_LOGCA_FIELD, OUT_GEOLOGY_LOGCA_COG, "geology_logca"),

        ("Geology_CaCoef", DO_GEOLOGY_CA_COE, GEOLOGY_CA_FC, GEOLOGY_CA_COE_FIELD, OUT_GEOLOGY_CA_COE_COG, "geology_cacoe"),

        ("Quaternary", DO_QUATERNARY, QUATERNARY_SHP, QUATERNARY_FIELD, OUT_QUATERNARY_COG, "quaternary")
    ]

    jobs = [j for j in all_jobs if j[1] and not should_skip(j[4])]

    msg("------------------------------------------------------------")
    msg(f"Planned outputs (maps to build): {len(jobs)}")

    if len(jobs) == 0:
        msg("Nothing to do (requested COGs already exist, or toggles off).")
        return

    msg("Will build: " + ", ".join([j[0] for j in jobs]))
    msg("------------------------------------------------------------")

    steps_per_map = 4
    total_steps = len(jobs) * steps_per_map

    arcpy.SetProgressor("step", "Rasterizing maps...", 0, total_steps, 1)

    scratch_gdb = arcpy.env.scratchGDB
    global_step = 0

    for name, _, fc, field, out, stage in jobs:

        process_vector_job(
            job_name=name,
            in_fc=fc,
            value_field=field,
            out_cog=out,
            stage_name=stage,
            mask_sr=mask_sr,
            mask_raster=MASK_RASTER,
            scratch_gdb=scratch_gdb,
            global_step=global_step
        )

        global_step += steps_per_map

    msg("------------------------------------------------------------")
    msg("Sanity check (dimensions):")

    msg(f"Mask rows/cols: {mask_ras.height} x {mask_ras.width}")

    for out in [
        OUT_MARINE_COG,
        OUT_LANDUSE_COG,
        OUT_GEOLOGY_LOGCA_COG,
        OUT_GEOLOGY_CA_COE_COG,
        OUT_QUATERNARY_COG
    ]:

        if os.path.exists(out):
            r = arcpy.Raster(out)
            msg(f"{os.path.basename(out)} rows/cols: {r.height} x {r.width}")

    arcpy.ResetProgressor()

    msg("Done.")
    msg("------------------------------------------------------------")


if __name__ == "__main__":
    main()
```

```python
###Euclidean distance....doesn't work

import arcpy
import os
import re
from arcpy.sa import EucDistance, ExtractByMask

# ============================================================
# USER INPUTS
# ============================================================

# Input feature class inside file geodatabase
input_polygons = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Geologi_0000_Norge_32633_Kalkinnhold_Berggrunn_FGDB\KalkinnholdBerggrunn.gdb\Kalkinnhold_berggrunnN250"

# Field with bedrock classes
class_field = "kalkinnhold_hovedbergart"

# Mask raster
mask_raster = r"E:\Alpha_earth\qc\alphaearth_mosaic_epsg25833_band1_qc_cog.tif"

# Output folder
output_folder = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Kalk_eclidean_dist"

# Constant field used for rasterization
constant_field = "RAST_VAL"

# ============================================================
# SETUP
# ============================================================

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")

# Validate inputs early
if not arcpy.Exists(input_polygons):
    raise ValueError(f"Input polygon dataset does not exist:\n{input_polygons}")

if not arcpy.Exists(mask_raster):
    raise ValueError(f"Mask raster does not exist:\n{mask_raster}")

# Create output folders
split_folder = os.path.join(output_folder, "01_split_polygons")
raster_folder = os.path.join(output_folder, "02_class_rasters")
distance_folder = os.path.join(output_folder, "03_distance_rasters")
cog_folder = os.path.join(output_folder, "04_cog_rasters")

for folder in [output_folder, split_folder, raster_folder, distance_folder, cog_folder]:
    os.makedirs(folder, exist_ok=True)

# Read mask properties
mask_desc = arcpy.Describe(mask_raster)
mask_sr = mask_desc.spatialReference

# Set environments to match mask raster
arcpy.env.snapRaster = mask_raster
arcpy.env.cellSize = mask_raster
arcpy.env.extent = mask_raster
arcpy.env.mask = None
arcpy.env.outputCoordinateSystem = mask_sr

# ============================================================
# HELPER FUNCTIONS
# ============================================================

def safe_name(value, prefix="class"):
    txt = str(value).strip()
    txt = re.sub(r"[^\w\-]+", "_", txt)
    txt = re.sub(r"_+", "_", txt)
    txt = txt.strip("_")
    if txt == "":
        txt = "unknown"
    return f"{prefix}_{txt}"

def get_unique_values(feature_class, field_name):
    values = set()
    with arcpy.da.SearchCursor(feature_class, [field_name]) as cursor:
        for row in cursor:
            if row[0] is not None:
                values.add(row[0])
    return sorted(values, key=lambda x: str(x))

def field_is_text(feature_class, field_name):
    fld = [f for f in arcpy.ListFields(feature_class) if f.name == field_name][0]
    return fld.type in ["String", "Guid"]

def make_sql_query(feature_class, field_name, value, text_field=True):
    field_delim = arcpy.AddFieldDelimiters(feature_class, field_name)
    if value is None:
        return f"{field_delim} IS NULL"
    if text_field:
        safe_val = str(value).replace("'", "''")
        return f"{field_delim} = '{safe_val}'"
    return f"{field_delim} = {value}"

# ============================================================
# STEP 1 - CHECK FIELD AND FIND UNIQUE CLASSES
# ============================================================

field_names = [f.name for f in arcpy.ListFields(input_polygons)]
if class_field not in field_names:
    raise ValueError(
        f"Field '{class_field}' not found.\nAvailable fields:\n" + "\n".join(field_names)
    )

print("Reading unique classes...")
classes = get_unique_values(input_polygons, class_field)

if len(classes) == 0:
    raise ValueError(f"No non-null values found in field '{class_field}'.")

print(f"Found {len(classes)} classes in '{class_field}':")
for c in classes:
    print(f"  - {c}")

text_field = field_is_text(input_polygons, class_field)

# ============================================================
# STEP 2 - ADD CONSTANT FIELD FOR RASTERIZATION
# ============================================================

if constant_field not in field_names:
    print(f"Adding field '{constant_field}'...")
    arcpy.management.AddField(input_polygons, constant_field, "SHORT")
    arcpy.management.CalculateField(input_polygons, constant_field, "1", "PYTHON3")
else:
    print(f"Field '{constant_field}' already exists.")

# Refresh field list
field_names = [f.name for f in arcpy.ListFields(input_polygons)]

# ============================================================
# STEP 3 - PROCESS EACH CLASS
# ============================================================

for class_value in classes:
    print("\n--------------------------------------------------")
    print(f"Processing class: {class_value}")

    class_name = safe_name(class_value, prefix="kalk")

    out_polygon = os.path.join(split_folder, f"{class_name}.shp")
    out_class_raster = os.path.join(raster_folder, f"{class_name}.tif")
    out_dist_raw = os.path.join(distance_folder, f"{class_name}_dist_raw.tif")
    out_dist_final = os.path.join(distance_folder, f"{class_name}_dist.tif")
    out_cog = os.path.join(cog_folder, f"{class_name}_dist_cog.tif")

    # Select polygons for this class
    where_clause = make_sql_query(input_polygons, class_field, class_value, text_field)
    print(f"Selecting polygons with query: {where_clause}")
    arcpy.analysis.Select(input_polygons, out_polygon, where_clause)

    n = int(arcpy.management.GetCount(out_polygon)[0])
    print(f"Number of polygons selected: {n}")

    if n == 0:
        print("No polygons found for this class. Skipping.")
        continue

    # Polygon to raster
    print("Rasterizing polygons...")
    arcpy.conversion.PolygonToRaster(
        in_features=out_polygon,
        value_field=constant_field,
        out_rasterdataset=out_class_raster,
        cell_assignment="CELL_CENTER",
        priority_field=constant_field,
        cellsize=mask_raster
    )

    # Euclidean distance across full extent
    print("Computing Euclidean distance...")
    euc = EucDistance(out_class_raster)
    euc.save(out_dist_raw)

    # Apply mask
    print("Applying land mask...")
    masked = ExtractByMask(out_dist_raw, mask_raster)
    masked.save(out_dist_final)

    # Delete temporary raw raster
    if arcpy.Exists(out_dist_raw):
        arcpy.management.Delete(out_dist_raw)

    # Export as COG
    print("Exporting COG...")
    arcpy.management.CopyRaster(
        in_raster=out_dist_final,
        out_rasterdataset=out_cog,
        format="COG"
    )

print("\nFinished successfully.")
arcpy.CheckInExtension("Spatial")
```

```python
###One hot encoding

import arcpy
import os
from arcpy.sa import ExtractByMask, Con, IsNull

# ============================================================
# USER INPUTS
# ============================================================

input_polygons = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Kalkinnhold_bergart_pol_clip.shp"
class_field = "kalkinnhol"

mask_raster = r"E:\Alpha_earth\qc\alphaearth_mosaic_epsg25833_band1_qc_cog.tif"

# ASCII-safe temp workspace
temp_root = r"C:\kalk_tmp"
temp_gdb = os.path.join(temp_root, "temp.gdb")

# Final output
output_folder = r"C:\Users\acosta_pedro\OneDrive - Norges geologiske undersøkelse\Geochemistry NGU_2025\kalk_prosjekt3.0\National\Geology\Kalk_one_hot_encoding"

# ============================================================
# SETUP
# ============================================================

arcpy.env.overwriteOutput = True
arcpy.CheckOutExtension("Spatial")
arcpy.ClearWorkspaceCache_management()

os.makedirs(temp_root, exist_ok=True)
os.makedirs(output_folder, exist_ok=True)

if not arcpy.Exists(temp_gdb):
    arcpy.management.CreateFileGDB(temp_root, "temp.gdb")

# ============================================================
# ENVIRONMENT SETTINGS
# ============================================================

mask_desc = arcpy.Describe(mask_raster)

arcpy.env.snapRaster = mask_raster
arcpy.env.cellSize = mask_raster
arcpy.env.extent = mask_raster
arcpy.env.outputCoordinateSystem = mask_desc.spatialReference

# ============================================================
# CLASSES TO ENCODE (ONLY 1–5)
# ============================================================

target_classes = [1, 2, 3, 4, 5]

# ============================================================
# LOOP THROUGH CLASSES
# ============================================================

for cls in target_classes:
    print(f"\nProcessing class: {cls}")

    temp_layer = f"lyr_class_{cls}"
    temp_raster = os.path.join(temp_gdb, f"ras_class_{cls}")
    temp_masked = os.path.join(temp_gdb, f"ras_class_{cls}_mask")

    final_cog = os.path.join(output_folder, f"kalk_class_{cls}_cog.tif")

    where_clause = f"{arcpy.AddFieldDelimiters(input_polygons, class_field)} = {cls}"

    # --------------------------------------------------------
    # STEP 1 — SELECT CURRENT CLASS
    # --------------------------------------------------------

    arcpy.management.MakeFeatureLayer(input_polygons, temp_layer, where_clause)

    count = int(arcpy.management.GetCount(temp_layer)[0])
    if count == 0:
        print(f"  No polygons found for class {cls}, skipping.")
        arcpy.management.Delete(temp_layer)
        continue

    print(f"  Selected {count} polygons.")

    # --------------------------------------------------------
    # STEP 2 — RASTERIZE CURRENT CLASS
    # --------------------------------------------------------

    print("  Rasterizing selected polygons...")

    arcpy.conversion.PolygonToRaster(
        in_features=temp_layer,
        value_field=class_field,
        out_rasterdataset=temp_raster,
        cell_assignment="CELL_CENTER",
        priority_field=class_field,
        cellsize=mask_raster
    )

    # --------------------------------------------------------
    # STEP 3 — CONVERT TO BINARY
    # --------------------------------------------------------

    print("  Converting to binary raster...")

    binary_raster = Con(IsNull(temp_raster), 0, 1)

    # --------------------------------------------------------
    # STEP 4 — APPLY MASK
    # --------------------------------------------------------

    print("  Applying mask...")

    masked = ExtractByMask(binary_raster, mask_raster)
    masked.save(temp_masked)

    # --------------------------------------------------------
    # STEP 5 — EXPORT COG ONLY
    # --------------------------------------------------------

    print("  Exporting COG...")

    try:
        arcpy.management.CopyRaster(
            in_raster=temp_masked,
            out_rasterdataset=final_cog,
            format="COG",
            pixel_type="8_BIT_UNSIGNED",
            compression="LZW"
        )
        print("  COG created successfully.")
    except Exception as e:
        print(f"  COG export failed: {e}")

    arcpy.management.Delete(temp_layer)

# ============================================================
# CLEANUP
# ============================================================

arcpy.ClearWorkspaceCache_management()
arcpy.CheckInExtension("Spatial")

print("\nDONE")
```
